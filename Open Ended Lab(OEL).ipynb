# Step 1: Load the Dataset

* The dataset is imported into the program using basic Python file-handling techniques.

* Data is accessed directly from the file without relying on complex tools or frameworks.

* Electricity consumption values are read sequentially, one line at a time.

* Each value is carefully stored in a list for further processing.

* This approach demonstrates how raw data can be manually extracted and organized.

* It helps build a clear understanding of fundamental data handling in Python.

```python
building_data = []
```

```python
with open("building_2.csv", "r") as file:
    next(file)  
    for line in file:
        values = line.strip().split(",")
        building_data.append(float(values[0]))  
```

# Step 2: Calculate Annual Consumption

* The annual electricity consumption is calculated by summing all hourly usage values.

* A loop is implemented to iterate through the entire dataset.

* Each hourly value is added to compute the total energy consumption.

* This process aggregates data across all time periods in the year.

* It provides the total electricity consumed by the building annually.

```python
annual_consumption = 0
```

```python
for value in building_data:
    annual_consumption += value
```

```python
print("Annual Electricity Consumption:", annual_consumption, "kWh")
```

# Step 3: Calculate Monthly Consumption

* The yearly electricity data is divided into monthly segments by grouping every 30 days (720 hours).

* Each monthly group contains hourly consumption values for that period.

* The values in each group are summed to calculate total monthly electricity usage.

* This method organizes the data into manageable time-based sections.

* It helps in identifying trends and variations in monthly energy consumption.

```python
monthly_consumption = []
```

```python
hours_per_month = 24 * 30
```

```python
for i in range(0, len(building_data), hours_per_month):
    month_sum = sum(building_data[i:i + hours_per_month])
    monthly_consumption.append(month_sum)
```

```python
print("Monthly Consumption:", monthly_consumption)
```

# Step 4: Calculate Daily Consumption

* Daily electricity consumption is calculated by summing each 24-hour block of data.

* Hourly values within the same day are grouped together for accurate analysis.

* The total for each group represents the building’s energy usage in one day.

* This method provides a clear view of daily consumption trends over time.

* It allows identification of patterns in high and low energy usage days.

* Such analysis helps in planning and optimizing daily energy management strategies.

```python
daily_consumption = []
```

```python
for i in range(0, len(building_data), 24):
    daily_total = sum(building_data[i:i+24])
    daily_consumption.append(daily_total)
```

```python
print("Daily Consumption (first 10 days):", daily_consumption[:10])
```

# Step 5: Calculate Seasonal Consumption

The year is divided into four seasons:

* Winter: December to February.

* Spring: March to May.

* Summer: June to August.

* Autumn: September to November.  

Each season’s energy consumption is calculated by adding the relevant monthly values.  
This helps in understanding how energy usage changes with seasons.

```python
season_hours = 24 * 30 * 3  
```

```python
winter = sum(building_data[0:season_hours])
```

```python
spring = sum(building_data[season_hours:2*season_hours])
```

```python
summer = sum(building_data[2*season_hours:3*season_hours])
```

```python
autumn = sum(building_data[3*season_hours:4*season_hours])
```

```python
print("Winter Consumption:", winter)
```

```python
print("Spring Consumption:", spring)
```

```python
print("Summer Consumption:", summer)
```

```python
print("Autumn Consumption:", autumn)
```

# Step 6: Weekday vs Weekend Consumption

* Daily electricity data is categorized into weekdays and weekends for analysis.

* Monday to Friday are considered as weekdays, while Saturday and Sunday are treated as weekends.

* Hourly consumption values are grouped accordingly to calculate total energy for each category.

* The total energy consumption of weekdays and weekends is computed separately.

* Comparing these totals highlights differences in electricity usage patterns.

* This analysis helps identify behavioral trends and variations between working days and weekends.

```python
weekday_consumption = 0
```

```python
weekend_consumption = 0
```

```python
for i in range(len(daily_consumption)):
    if (i % 7) < 5:   # Monday to Friday
        weekday_consumption += daily_consumption[i]
    else:             # Saturday and Sunday
        weekend_consumption += daily_consumption[i]
```

```python
print("Weekday Consumption:", weekday_consumption)
```

```python
print("Weekend Consumption:", weekend_consumption)
```

# Interpretation

* The analysis of electricity consumption reveals clear variations over different time periods.

* Daily, monthly, and seasonal patterns highlight fluctuations in energy usage.

* Higher electricity consumption is observed during specific seasons.

* Weekdays generally show greater energy usage compared to weekends.

* Peaks and troughs in consumption indicate behavioral and operational trends.

* Understanding these patterns helps in predicting future energy demands.

* This analysis supports more efficient planning and management of building energy usage.

```python
# Monthly consumption interpretation
print("\n--- Monthly Consumption ---")
for i, month in enumerate(monthly_consumption, start=1):
    print(f"Month {i}: {month:.2f} kWh")
```

```python
# Find highest and lowest months
highest_month = max(monthly_consumption)
lowest_month = min(monthly_consumption)
print(f"Highest consumption month: {monthly_consumption.index(highest_month)+1} ({highest_month:.2f} kWh)")
print(f"Lowest consumption month: {monthly_consumption.index(lowest_month)+1} ({lowest_month:.2f} kWh)")
```

```python
# Seasonal consumption interpretation
print("\n--- Seasonal Consumption ---")
seasons = ["Winter", "Spring", "Summer", "Autumn"]
season_values = [winter, spring, summer, autumn]
for s, val in zip(seasons, season_values):
    print(f"{s}: {val:.2f} kWh")
```

```python
# Weekday vs Weekend comparison
print("\n--- Weekday vs Weekend Consumption ---")
print(f"Total Weekday Consumption: {weekday_consumption:.2f} kWh")
print(f"Total Weekend Consumption: {weekend_consumption:.2f} kWh")
```

```python
if weekday_consumption > weekend_consumption:
    print("Weekdays consume more electricity than weekends.")
else:
    print("Weekends consume more electricity than weekdays.")
```
